```python
import pandas as pd

df = pd.read_csv("../Data/Raw/online_retail_II.csv")

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)
```

```python
print("Column names:")
print(df.columns.tolist())

print("\nFirst 5 rows:")
display(df.head())
```

```python
print("DATA TYPES")
print(df.dtypes)

print("\nMISSING VALUES")
print(df.isnull().sum())

print("\nDUPLICATE ROWS")
print(df.duplicated().sum())
```

```python
# Convert InvoiceDate to datetime
df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"])

# Basic dataset timeline
print("DATASET DATE RANGE")
print("Start date:", df["InvoiceDate"].min())
print("End date:", df["InvoiceDate"].max())

# Customer validation
print("\nUNIQUE CUSTOMERS")
print(df["Customer ID"].nunique())

# Invoice validation
print("\nUNIQUE INVOICES")
print(df["Invoice"].nunique())

# Missing Customer ID percentage
missing_customer_pct = df["Customer ID"].isnull().mean() * 100

print("\nMISSING CUSTOMER ID")
print(f"{missing_customer_pct:.2f}%")
```

```python
# Investigate duplicate rows

duplicate_rows = df[df.duplicated()]

print("Number of duplicate rows:")
print(duplicate_rows.shape[0])

display(duplicate_rows.head(10))
```

```python
# Create a working copy
clean_df = df.copy()

# Remove exact duplicate rows
clean_df = clean_df.drop_duplicates()

# Remove rows without a Customer ID
clean_df = clean_df.dropna(subset=["Customer ID"])

# Convert InvoiceDate to datetime
clean_df["InvoiceDate"] = pd.to_datetime(clean_df["InvoiceDate"])

# Remove cancelled invoices
clean_df = clean_df[~clean_df["Invoice"].str.startswith("C")]

# Keep only positive quantities
clean_df = clean_df[clean_df["Quantity"] > 0]

# Keep only positive prices
clean_df = clean_df[clean_df["Price"] > 0]

print("Original dataset shape:", df.shape)
print("Cleaned dataset shape:", clean_df.shape)
```

```python
# Save cleaned dataset

clean_df.to_csv(
    "../Data/Processed/online_retail_cleaned.csv",
    index=False
)

print("Cleaned dataset saved successfully!")
```

```python
# Create revenue column

clean_df["Revenue"] = clean_df["Quantity"] * clean_df["Price"]

print("Revenue column created!")
display(clean_df.head())
```

```python
# Basic business metrics

total_revenue = clean_df["Revenue"].sum()

total_customers = clean_df["Customer ID"].nunique()

total_invoices = clean_df["Invoice"].nunique()

average_order_value = total_revenue / total_invoices

print("BUSINESS OVERVIEW")
print("----------------------")

print(f"Total Revenue: £{total_revenue:,.2f}")
print(f"Total Customers: {total_customers:,}")
print(f"Total Orders: {total_invoices:,}")
print(f"Average Order Value: £{average_order_value:,.2f}")
```

```python
# Calculate number of orders per customer

customer_orders = (
    clean_df
    .groupby("Customer ID")["Invoice"]
    .nunique()
)

repeat_customers = (customer_orders >= 2).sum()

one_time_customers = (customer_orders == 1).sum()

total_customers = clean_df["Customer ID"].nunique()

repeat_rate = repeat_customers / total_customers * 100

print("CUSTOMER BEHAVIOR")
print("----------------------")

print(f"Total Customers: {total_customers:,}")
print(f"Repeat Customers: {repeat_customers:,}")
print(f"One-Time Customers: {one_time_customers:,}")
print(f"Repeat Customer Rate: {repeat_rate:.2f}%")
```
